```python
from pathlib import Path
import pydicom

dicom_dir = Path("sourcedata/dicoms_uw/0ACSF02M/1401/DICOM")

datasets = []
for f in sorted(dicom_dir.glob("*.dcm")):
    ds = pydicom.dcmread(f, stop_before_pixels=True)
    datasets.append(ds)

print(f"Read {len(datasets)} DICOM files")
```

```python
for i, ds in enumerate(datasets[:]):
    print(
        f"datasets[{i}]: "
        f"Phase Encode={ds[(0x2005, 0x1564)].value}, "
        f"AcquisitionTime={ds.get('AcquisitionTime')}, "
        f"InstanceCreationTime={ds.get('InstanceCreationTime')}"
    )
```
